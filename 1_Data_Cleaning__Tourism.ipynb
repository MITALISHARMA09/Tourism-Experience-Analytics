### 7. User dataset

```python
df_user=pd.read_excel(f"{data_path}/User.xlsx")
df_user.head()
```

```python
df_user.info()
```

Why CityId is float64 (important insight ) ?

CityId became float64 only because of NaN.

Pandas rule:

- Integer column + NaN → converted to float

So this is not a data design issue, just a missing-value side effect.

```python
df_user.describe()
```

```python
df_user.isnull().sum()
```

```python
df_user[df_user["CityId"].isna()]
```

```python
df_user = df_user.dropna(subset=["CityId"]).copy()
df_user['CityId'] = df_user['CityId'].astype(int)
```

“The user dataset had 4 missing CityId values (0.01%). Since they were negligible and not critical, those rows were removed to preserve data integrity and consistent data types.”

```python
df_user.duplicated().sum()
```
